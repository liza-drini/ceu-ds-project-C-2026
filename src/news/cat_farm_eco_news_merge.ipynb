# Final Merge: Farm, Cat, Economic, and News Data

This notebook creates the final merged dataset used in the project by combining the national-level farm, cat, economic, and news data.

Because the news dataset is available only for Australia as a whole and covers a shorter period, the merge is restricted to **All Australia** observations from **2017 onward**.

The goal of this notebook is to:

1. load the prepared news dataset and the merged farm-cat-economic dataset;
2. keep only national-level observations;
3. restrict the data to the period covered by the news variables;
4. merge the datasets by year;
5. save the final result as `farm_cat_eco_news.csv`.

```python
import pandas as pd

news_df = pd.read_csv('../datasets/news.csv')
farm_cat_eco_merged = pd.read_csv("../datasets/farm_cat_eco.csv")
```

```python
# Keep only national-level observations from 2017 onward
# This matches the coverage of the news dataset
aa = farm_cat_eco_merged[
    (farm_cat_eco_merged["territory"] == "All Australia") &
    (farm_cat_eco_merged["year"] >= 2017)
].copy()

# Merge the national farm-cat-economic data with the news data by year
farm_cat_eco_news_merged = aa.merge(news_df, on="year", how="left")
```

```python
farm_cat_eco_news_merged.to_csv("../datasets/farm_cat_eco_news.csv", index=False)
```
