```python
## Checking those rows where missing values

cols_to_check = [
    'Location',
    'Details',
    'Cost_estimate_per_year_2017_USD_exchange_rate',
    'Cost_estimate_per_year_2017_USD_PPP',
    'State|Province|Administrative_area'
]

rows_with_missing = eco_df[
    eco_df[cols_to_check].isna().any(axis=1)
]


rows_with_missing[
    cols_to_check + [
        'Official_country',
        'State|Province|Administrative_area',
        'Details',
        'Type_of_cost_merged'
    ]
]


rows_with_missing


 
```

```python
## Drop 3 inconsistent rows (where the cost is nan and can influence on analysis)
eco_df = eco_df.dropna(
    subset=['Cost_estimate_per_year_2017_USD_PPP']
)
```

```python
## Check whther only 3 rows were dropped
len(eco_df)
```

```python
## In obj types of columns change Nan to Unspecified
eco_df[['State|Province|Administrative_area', 'Location', 'Details']] = (
    eco_df[['State|Province|Administrative_area', 'Location', 'Details']]
    .fillna('Unspecified')
)

```

```python
## Eliminate the rows that are connected to prey

kill_hunt_keywords = [
   'hunting',
   'hunt',
   'predation',
   'kill',
   'kills'
]

pattern = '|'.join(kill_hunt_keywords)

eco_df = eco_df[
    ~eco_df['Details']
        .str.lower()
        .str.contains(pattern, regex=True, na=False)
]

len(eco_df)
```

```python
## Check
predation_keywords = [
   'hunt',
   'predation',
   'kill',
   'kills'
   ]

pattern = '|'.join(predation_keywords)

predation_rows = eco_df[
    eco_df['Details']
        .str.lower()
        .str.contains(pattern, regex=True, na=False)
]

predation_rows[['Details']].head(20)
```

```python
eco_df
```

```python
## Raws that are related to Toxoplasmosis only

tox_keywords = [
    'toxoplasma',
    'toxoplasmosis',
    'toxoplasma gondii'
]

df_toxo_eco = eco_df[
    eco_df['Details']
    .str.lower()
    .str.contains('|'.join(tox_keywords), na=False)
]
```

```python
df_toxo_eco
```

```python
## CATS only

cats_only_df = eco_df[
    eco_df['Common_name']
    .astype(str)
    .str.lower()
    .str.contains(r'\bcat\b', na=False) &
    ~eco_df['Common_name'].str.contains(r',|\+|/|;|\band\b', case=False, na=False)
]

cats_only_df
```

```python
cats_only_df['Common_name'].value_counts()
```

```python
## Subset with multispecies
multi_species_df_eco = eco_df[
    eco_df['Common_name']
    .str.lower()
    .str.contains('cat', na=False) &
    eco_df['Common_name']
    .str.contains(',|\+| and |;', na=False)
]

multi_species_df_eco
```
